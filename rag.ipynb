###SPLITTING DATA ###

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitters = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50,

)
chunks = text_splitters.split_documents(documents)
print(chunks)
```

EMBEDED INTO DATA

```python
embeddings_model = JinaEmbeddings(model_name ="jina-embeddings-v5-text-small")
print("Embedded model is ready with JinaEmbeddings")
```

STORE DATA IN VECTOR DATABASE

```python
from langchain_community.vectorstores import FAISS

vector_store = FAISS.from_documents(chunks, embeddings_model)

print("CHUNKS ARE STORED", vector_store.index.ntotal)
```

```python
####SIMILARITY SEARCH######

test_query = "what is dengue"

top_matches = vector_store.similarity_search(test_query, k=2)
print(f"Query: {test_query}\n")
for i,match in enumerate(top_matches,start = 1):
    print(f"--- Match {i} ---")
    print(match.page_content)
    print()
```

```python
def ask_MEDIQUEST (question: str) -> str:
    """Send a question to the RAG agent and print a nicely formatted answer."""

    print("=" * 60)
    print("QUESTION:", question)
    print("-" * 60)

    response = MEDIQUEST.invoke({"messages": [{"role": "user", "content": question}]})
    answer = response["messages"][-1].content

    print("ANSWER:", answer)
    print("=" * 60)
    print()

    return answer
```

TOOL

```python
retriever = vector_store.as_retriever(
    search_kwargs={"k": 3}  # Retrieve the top 3 relevant chunks
)

def search_MEDIQUEST(question: str) -> str:
    '''
    Search for disease and disorders
    '''
    matching_chunks = retriever.invoke(question)

    for chunk in matching_chunks:
        print("Content:", chunk.page_content)
        print("Metadata:", chunk.metadata)
        print("-" * 50)

    return "\n".join(chunk.page_content for chunk in matching_chunks)
```
